```python
import pandas as pd
data= {
    "student_id":[101, 102, 103, 104, 105, 106, 102, 107],
    "Student": ["asha", "ravi", "priya", "john", "anu", "kiran", "ravi", "meena"],
    "department":["cse", "cse", "ece", "ece", "cse", "eee", "cse", "eee"],
    "attendance":[90, 75, 95, 60, 80, 70, 75, 85],
    "internal_marks":[85, 70, 90, 55, 78, 65, 70, None],
    "final_marks":[88, 72,92, 50, 80, 68, 72, 82]
}
```

```python
df= pd.DataFrame(data)
df.to_csv("students.csv", index=False)
print("students.csv created successfully!")
```

```python
df= pd.read_csv("students.csv")
print("Original Dataset:")
print(df)
```

```python
print("\nFirst Five Records:")
print(df.head())
```

```python
print("\nLast Five Records:")
print(df.tail())
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("\nNumber of duplicate records:")
print(df.duplicated().sum())
```

```python
print("\nAverage final marks by department:")
print(df.groupby("department")["final_marks"].mean())
```

```python
courses = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106,107],
    "course":["python", "python", "java", "python", "java", "python", "java"]
})
```

```python
merged_df=pd.merge(
    df,
    courses,
    on="student_id"
)
print("\nMerged Dataset:")
print(merged_df)
```
